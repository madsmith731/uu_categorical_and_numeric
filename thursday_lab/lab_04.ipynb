### Q1

```python
#setting up grid of K=10 equally spaced values between 0 and 1
grid_values = np.linspace(0,1,10)

#choosing a point from the grid
rng = np.random.default_rng(seed=100)
rng.choice(grid_values, size=1)
```

```python
#repeating above steps 5,000 times
rng = np.random.default_rng(seed=100)
results_q1 = rng.choice(grid_values, size=(1,5000))
```

```python
#creating ecdf plot
sns.ecdfplot(data=results_q1.flatten())
```

```python
#re-running with K=50
grid_values = np.linspace(0,1,50)
rng = np.random.default_rng(seed=100)
results_q1 = rng.choice(grid_values, size=(1,5000))
sns.ecdfplot(data=results_q1.flatten())
```

As I increased the value of K, the distribution approached a uniform distribution. This is because we spread the probability of eeach number appearing evenly across the range of values from 0 to 1.

### Q2

```python
#drawing 32 values from Uniform[-sqrt(3), sqrt(3)
rng = np.random.default_rng(seed=100)
n_values = rng.uniform(-np.sqrt(3), np.sqrt(3), size=32)

#calculating mean and multiplying by sqrt(n)
np.mean(n_values) * np.sqrt(32)
```

```python
#repeating 5,000 times
rng = np.random.default_rng(seed=100)

num_times = 5000
n = 32

result_q2 = []
for i in range(5000):
    gen_list = rng.uniform(-np.sqrt(3), np.sqrt(3), size=n)
    mean_sqrt = np.mean(gen_list) * np.sqrt(n)
    result_q2.append(mean_sqrt)

np.array(result_q2)
```

```python
sns.ecdfplot(data=result_q2)
```

```python
#re-running with n=100
rng = np.random.default_rng(seed=100)

num_times = 5000
n = 100

result_q2 = []
for i in range(5000):
    gen_list = rng.uniform(-np.sqrt(3), np.sqrt(3), size=n)
    mean_sqrt = np.mean(gen_list) * np.sqrt(n)
    result_q2.append(mean_sqrt)

np.array(result_q2)

sns.ecdfplot(data=result_q2)
```

```python
#re-running with n=1000
rng = np.random.default_rng(seed=100)

num_times = 5000
n = 1000

result_q2 = []
for i in range(5000):
    gen_list = rng.uniform(-np.sqrt(3), np.sqrt(3), size=n)
    mean_sqrt = np.mean(gen_list) * np.sqrt(n)
    result_q2.append(mean_sqrt)

np.array(result_q2)

sns.ecdfplot(data=result_q2)
```

As n increased, this simulation appeared to approach a normal distribution. This is because the ecdf plots above show a distribution which is symatrical around zero, which is our expected mean of numbers randomly generated between -sqrt(3) and +sqrt(3), with the steepest slopes falling around zero and the smallest slopes farthest away from zero.
