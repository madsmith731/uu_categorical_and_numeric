### Q2

```python
#drawing 32 values from Uniform[-sqrt(3), sqrt(3)]
rng = np.random.default_rng(seed=100)
n_values = rng.uniform(-np.sqrt(3), np.sqrt(3), size=32)

#calculating mean and multiplying by sqrt(n)
np.mean(n_values) * np.sqrt(32)
```

```python
#repeating 5,000 times
rng = np.random.default_rng(seed=100)

num_times = 5000
n = 32

result_q2 = []
for i in range(5000):
    gen_list = rng.uniform(-np.sqrt(3), np.sqrt(3), size=n)
    mean_sqrt = np.mean(gen_list) * np.sqrt(n)
    result_q2.append(mean_sqrt)

np.array(result_q2)
```

```python
sns.ecdfplot(data=result_q2)
plt.title('Q2 Uniform[-sqrt(3), sqrt(3)]: n=32')
```

```python
#re-running with n=1000
rng = np.random.default_rng(seed=100)

num_times = 5000
n = 1000

result_q2 = []
for i in range(5000):
    gen_list = rng.uniform(-np.sqrt(3), np.sqrt(3), size=n)
    mean_sqrt = np.mean(gen_list) * np.sqrt(n)
    result_q2.append(mean_sqrt)

np.array(result_q2)

sns.ecdfplot(data=result_q2)
plt.title('Q2 Uniform[-sqrt(3), sqrt(3)]: n=1,000')
```

As n increased, this simulation appeared to approach a normal distribution. This is because the ecdf plots above show a distribution which is symmetrical around zero, which is the expected mean of numbers randomly generated between -sqrt(3) and +sqrt(3). The steepest slopes falling around zero and the smallest slopes farthest away from zero.

### Q3

```python
#defining p
p = 0.5

#defining outcomes
coin_sides = np.array([
    "H",
    "T"
])

#defining probabilities of outcomes
probabilities = np.array([
    p,
    1-p
])

rng = np.random.default_rng(seed=100)
coin_flips = rng.choice(coin_sides, size = 40, p = probabilities)

total_heads = np.sum(coin_flips == "H")

(total_heads - 40*p) / np.sqrt( 40 * p * (1-p) )
```

```python
#repeating above code 5_000 times
rng = np.random.default_rng(seed=100)

b = 5_000
n = 40

results_q3 = []
for i in range(b):
    #making n choices based on coin_sides and probabilities
    coin_flips = rng.choice(coin_sides, size = (n), p = probabilities)
    #counting the number of "H"
    total_heads = np.sum(coin_flips == "H")
    #subtracting expected value of heads from total and dividing by sqrt( n * p * (1-p) )
    final = (total_heads - n*p) / np.sqrt( n * p * (1-p) )
    #appending calculation to a list
    results_q3.append(final)

sns.ecdfplot(data=results_q3)
plt.title('Q3 Distribution of the Survivors: n=40')
```

```python
#repeating above code for n=100
rng = np.random.default_rng(seed=100)


b = 5_000
n = 100

results_q3 = []
for i in range(b):
    coin_flips = rng.choice(coin_sides, size = (n), p = probabilities)

    total_heads = np.sum(coin_flips == "H")

    final = (total_heads - n*p) / np.sqrt( n * p * (1-p) )

    results_q3.append(final)

sns.ecdfplot(data=results_q3)
plt.title('Q3 Distribution of the Survivors: n=100')
```

```python
#repeating above code for n=1000
rng = np.random.default_rng(seed=100)


b = 5_000
n = 1_000

results_q3 = []
for i in range(b):
    coin_flips = rng.choice(coin_sides, size = (n), p = probabilities)

    total_heads = np.sum(coin_flips == "H")

    final = (total_heads - n*p) / np.sqrt( n * p * (1-p) )

    results_q3.append(final)

sns.ecdfplot(data=results_q3)
plt.title('Q3 Distribution of the Survivors: n=1,000')
```

This approaches a normal distribution as n increases. This makes sense because it's expected that when randomly choosing a value with consistent probability that most of the time it would result in a number of heads close to our expected value, giving a distribution centered around zero.
